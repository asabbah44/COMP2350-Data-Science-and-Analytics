## 1. Brief concept reminder

**Data science** uses data, careful reasoning, and suitable tools to understand a problem and communicate useful evidence.

Important terms for this week:

- **Problem:** the situation that needs to be understood or improved.
- **Question:** a focused statement of what we want to learn.
- **Population:** the full group we want to understand.
- **Observational unit:** one person, object, place, event, or time period described by one record.
- **Variable:** a characteristic recorded for each observational unit.
- **Outcome variable:** the main result we want to understand.
- **Explanatory variable:** information that may help explain differences in the outcome.
- **Evidence:** information from data that supports or weakens a claim.
- **Association:** two variables change together in the observed data. Association alone does not prove that one variable causes the other.

A good data question identifies the population, the outcome, and the information needed to investigate the problem.


## 2. A simple data-science workflow

For this course, begin with the following sequence:

**Problem → Question → Data needed → Variables → Evidence → Interpretation → Communication**

1. **Understand the problem.** Why does it matter, and who is affected?
2. **Formulate a question.** Make it focused and answerable with data.
3. **Identify the data needed.** Decide what or who should be observed and what should be recorded.
4. **Examine the evidence.** Look for information that supports or weakens possible explanations.
5. **Interpret carefully.** Consider other explanations and limitations.
6. **Communicate.** Explain the finding in clear language for the intended audience.

Later weeks introduce Python and other tools for carrying out these stages. The problem and question still come first.


## 3. Tools and roles

Data-science work usually combines several contributions:

- A **subject expert** explains the real-world context and what a useful answer would mean.
- A **data analyst or data scientist** organizes questions, examines data, and evaluates evidence.
- A **decision-maker** uses the findings while considering practical and ethical consequences.
- **Computing tools** help store, organize, analyze, and present data. A tool does not choose a good question or guarantee a correct conclusion.

One person may perform more than one role, but all roles require clear communication and responsible use of data.


## 4. Guided example

### General problem

Some students are not achieving the course results they expect.

### Focused question

**Among students taking the course this semester, is weekly study time associated with final course performance?**

### Reasoning through the question

- **Population:** students taking the course this semester.
- **Observational unit:** one student.
- **Outcome variable:** final course score.
- **Possible explanatory variable:** weekly study time.
- **Other information that may matter:** attendance, prior preparation, sleep, access to learning resources, and motivation.
- **Data needed:** for each participating student, record the relevant information using clear definitions and consistent units.
- **Useful evidence:** a clear pattern showing whether students who report different study times also tend to have different final scores.
- **Important limitation:** even if study time and final score are associated, other factors may help explain the pattern. The data would not automatically prove that study time caused the score difference.

Notice that the example begins with a problem and a question—not with software.
