## Setup

Run the next cell once. No editing is required.


```python
from IPython import get_ipython
from IPython.display import Markdown, display

get_ipython().run_line_magic("matplotlib", "inline")
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

def show_table(headers, rows):
    header = "| " + " | ".join(headers) + " |"
    separator = "|" + "|".join(["---"] * len(headers)) + "|"
    body = ["| " + " | ".join(str(value) for value in row) + " |" for row in rows]
    display(Markdown("\n".join([header, separator, *body])))

display(Markdown("**Notebook ready.** Continue by running each demonstration cell."))
```

## Demonstration 1 — Data science, analytics, and analytical goals

Presentation 1 defines data science as learning about the world from data using computation. Data analytics emphasizes understanding what happened and why. Run the cell to compare the three introductory data-science goals and see where data creates value.


```python
display(Markdown("### Exploration, inference, and prediction"))
show_table(
    ["Goal", "Purpose", "Example"],
    [
        ["Exploration", "Discover patterns, structure, trends, and anomalies in observed data", "Find the highest-traffic locations in a city"],
        ["Inference", "Use a sample to draw conclusions about a wider population or process", "Use a student sample to study the wider student population"],
        ["Prediction", "Use observed features to estimate an unknown or future outcome", "Estimate the price of a new property"],
    ],
)

display(Markdown("### Examples of value"))
show_table(
    ["Domain", "Available evidence", "Possible use"],
    [
        ["Healthcare", "Patient history and service demand", "Monitor risk and forecast demand"],
        ["Mobility", "Traffic, routes, and travel demand", "Understand congestion and improve planning"],
        ["Retail", "Transactions and customer activity", "Understand demand and customer churn"],
        ["Cybersecurity", "Authentication logs and network events", "Detect unusual patterns and prioritize alerts"],
    ],
)
```

**Interpretation:** The analytical goal depends on the question. The same project may use exploration, inference, and prediction at different stages.


## Demonstration 2 — From a vague problem to a stronger data question

Run the cell to see how the concern “Customers are leaving” becomes a question that defines the outcome, population, unit of analysis, time frame, and evidence needed.

## Demonstration 3 — Association is not the same as causation

The presentation uses ice-cream sales and swimming incidents to show how two quantities can rise together without one directly causing the other. Run the cell and inspect the two plots.


```python
temperature = [20, 24, 28, 32, 36]
ice_cream_sales = [30, 45, 65, 85, 105]
swimming_incidents = [2, 3, 5, 7, 9]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(temperature, ice_cream_sales, marker="o", color="#2f6b9a")
axes[0].set_title("Ice-cream sales and temperature")
axes[0].set_xlabel("Temperature (°C)")
axes[0].set_ylabel("Ice-cream sales index")

axes[1].plot(temperature, swimming_incidents, marker="o", color="#c45a3c")
axes[1].set_title("Swimming incidents and temperature")
axes[1].set_xlabel("Temperature (°C)")
axes[1].set_ylabel("Number of incidents")

plt.tight_layout()
plt.show()

display(Markdown(
    "**Observed association:** both quantities increase as temperature rises.  \n"
    "**Conclusion allowed:** the variables move together in these observations.  \n"
    "**Conclusion not allowed:** buying ice cream causes swimming incidents.  \n"
    "**Alternative explanation:** hot weather can influence both quantities."
))
```

**Interpretation:** Association can support a cautious conclusion. A causal claim requires stronger evidence and an appropriate study design.
